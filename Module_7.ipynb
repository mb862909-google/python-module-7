# **Question 2.  How does broadcasting work in NumPy ?**
Answer:

Broadcasting in NumPy is a powerful mechanism that allows arrays of different shapes to be used together in arithmetic operations without explicitly resizing or copying data.

It automatically “stretches” smaller arrays along compatible dimensions so they match the shape of the larger one — allowing fast, vectorized operations.

🔹 Basic Idea

When performing operations (like addition, subtraction, etc.) on two arrays:

1. NumPy compares their shapes element-wise (from right to left).

2. Two dimensions are compatible if:

    * They are equal, or

    * One of them is 1.

3. If dimensions are compatible, NumPy broadcasts (repeats) the smaller array along that dimension.

🔹 Example 1: Simple Broadcasting

```python
import numpy as np

a = np.array([1, 2, 3])
b = 5

result = a + b
print(result)
```

✅ Here, b (a scalar) is broadcast across all elements of a.

## **🔹 Example 2: Different Shapes**

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])
b = np.array([10, 20, 30])

result = a + b
print(result)
```

✅ NumPy broadcasts b along the first dimension (rows) to match a’s shape (2, 3).

## **🔹 Example 3: 2D + 1D (Column Broadcasting)**

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])
b = np.array([[10],
              [20]])
result = a + b
print(result)
```

✅ b is broadcast across columns.

## **🔹 Broadcasting Rules Summary**

| Condition               | Description                                |
| ----------------------- | ------------------------------------------ |
| Same shape              | Operates element-wise                      |
| One dimension is 1      | That array is stretched to match the other |
| Incompatible dimensions | Raises an error                            |

# **Question 4  Explain the use of the groupby() method in Pandas?**

Answer:

The groupby() method in Pandas is used to split data into groups based on one or more columns, and then apply operations (like aggregation, transformation, or filtering) to each group independently.

It’s a very powerful tool for data summarization and analysis, especially when you need to perform operations like “find the average sales per region” or “total revenue by product category”.

## **🔹 The GroupBy Process**

The groupby() method follows a “split–apply–combine” approach:

1. Split – Divide the DataFrame into groups based on some criteria (like a column’s values).

2. Apply – Perform a function on each group (e.g., sum, mean, count).

3. Combine – Merge the results back into a new DataFrame or Series.

## **🔹 Example: Basic Usage**

```python
import pandas as pd

data = {
    'Department': ['Sales', 'IT', 'Sales', 'HR', 'IT', 'HR'],
    'Employee': ['Aakash', 'Riya', 'Meera', 'Suresh', 'Neha', 'Prakash'],
    'Salary': [50000, 60000, 55000, 45000, 65000, 47000]
}

df = pd.DataFrame(data)
print(df)
```

## **🔹 1. Grouping and Aggregating**

```python
grouped = df.groupby('Department')['Salary'].mean()
print(grouped)
```

✅ Calculates the average salary per department.

## **🔹 2. Multiple Aggregations**

```python
result = df.groupby('Department')['Salary'].agg(['mean', 'sum', 'count'])
print(result)
```

## **🔹 3. Group by Multiple Columns**

```python
df.groupby(['Department', 'Employee'])['Salary'].sum()
```

## **🔹 4. Applying Custom Functions**

```python
def range_func(x):
    return x.max() - x.min()

df.groupby('Department')['Salary'].apply(range_func)
```

✅ Returns the salary range within each department.

## **🔹 5. Iterating Through Groups**

```python
for dept, group in df.groupby('Department'):
    print(dept)
    print(group)
```

## **Summary**

| Step        | Description                               |
| ----------- | ----------------------------------------- |
| **Split**   | Divide data into groups based on a column |
| **Apply**   | Perform aggregation or transformation     |
| **Combine** | Merge the results into a DataFrame        |

# **Question 6.  What are the differences between NumPy arrays and Python lists ?**

Answer:

While NumPy arrays and Python lists may look similar (both store collections of elements), they are fundamentally different in how they store, process, and handle data — especially for numerical and scientific computing.

## *🔹 1. Data Type*

| Feature       | **NumPy Array**                                     | **Python List**                                            |
| ------------- | --------------------------------------------------- | ---------------------------------------------------------- |
| **Data Type** | Stores elements of **the same type** (homogeneous). | Can store elements of **different types** (heterogeneous). |
| **Example**   | `[1, 2, 3]` (all integers)                          | `[1, 'two', 3.0]` (mixed types allowed)                    |


✅ NumPy arrays are type-consistent, which makes operations faster and more memory-efficient

## **🔹 2. Performance (Speed)**

|                 |                                                                                                             |
| --------------- | ----------------------------------------------------------------------------------------------------------- |
| **NumPy Array** | Much **faster** because it uses **C-based** implementation and performs operations in bulk (vectorization). |
| **Python List** | Slower — operations happen element by element using Python loops.                                           |


Example:

```python
import numpy as np, time

arr = np.arange(1000000)
lst = list(range(1000000))

start = time.time(); arr * 2; print("NumPy:", time.time() - start)
start = time.time(); [x*2 for x in lst]; print("List:", time.time() - start)
```

✅ NumPy will be significantly faster.


## **🔹 3. Memory Efficiency**

|                 |                                                          |
| --------------- | -------------------------------------------------------- |
| **NumPy Array** | Uses **less memory** (stores data in contiguous blocks). |
| **Python List** | Stores **references** to objects (more overhead).        |


✅ NumPy arrays are compact and better for large datasets.

## **🔹 4. Functionality**

| Feature                 | **NumPy Array**                               | **Python List**                                          |
| ----------------------- | --------------------------------------------- | -------------------------------------------------------- |
| Mathematical operations | Supported directly (`arr1 + arr2`, `arr * 5`) | Not supported (`[1,2,3] + [4,5,6]` concatenates instead) |
| Broadcasting            | Yes                                           | No                                                       |
| Statistical methods     | Built-in (`np.mean`, `np.sum`, `np.std`)      | Must be implemented manually                             |


## **🔹 5. Dimensionality**

|                 |                                                                          |
| --------------- | ------------------------------------------------------------------------ |
| **NumPy Array** | Supports **multi-dimensional** arrays (2D, 3D, etc.).                    |
| **Python List** | Only 1D lists; nested lists are used for 2D structures (less efficient). |

Example:

```python
np_array = np.array([[1, 2], [3, 4]])  # 2D array
list_2d = [[1, 2], [3, 4]]             # Nested list
```

## **🔹 6. Ease of Computation**

NumPy provides vectorized operations, meaning:

```python
a = np.array([1, 2, 3])
print(a * 2)
```

Whereas with lists:

```python
a = [1, 2, 3]
print([x * 2 for x in a])  # Requires looping
```

## **🔹 7. Use Cases**

| Use Case                                | Preferred     |
| --------------------------------------- | ------------- |
| Numerical / Scientific computing        | ✅ NumPy array |
| Simple data storage / small collections | ✅ Python list |

## **Question 8  What does the term “vectorized operation” mean in NumPy ?**

Answer:

In NumPy, a vectorized operation refers to performing mathematical or logical operations on entire arrays (vectors, matrices, etc.) without using explicit Python loops.

It allows computations to be fast, concise, and efficient, leveraging low-level C and Fortran optimizations under the hood.

🔹 Definition

A vectorized operation is an operation that applies a function element-wise to all elements of an array in one step, rather than iterating through elements with a loop.

🔹 Example: Without vs. With Vectorization
🐢 Without Vectorization (Using a Loop)

```python
data = [1, 2, 3, 4, 5]
result = []

for x in data:
    result.append(x * 2)

print(result)
```

### **⚡ With Vectorization (Using NumPy)**

```python
import numpy as np

data = np.array([1, 2, 3, 4, 5])
result = data * 2

print(result)
```

### **🔹 Why Vectorized Operations Are Powerful**

| **Feature**           | **Explanation**                                                                         |
| --------------------- | --------------------------------------------------------------------------------------- |
| **Speed**             | Uses optimized C code — much faster than Python loops.                                  |
| **Simplicity**        | Cleaner, more readable syntax.                                                          |
| **Memory Efficiency** | Operates on contiguous memory blocks.                                                   |
| **Parallelization**   | Executes multiple operations simultaneously (SIMD – Single Instruction, Multiple Data). |


### **🔹 Examples of Vectorized Operations**

| Operation                  | Example        | Result                           |
| -------------------------- | -------------- | -------------------------------- |
| **Addition**               | `a + b`        | Element-wise sum                 |
| **Subtraction**            | `a - b`        | Element-wise difference          |
| **Multiplication**         | `a * 2`        | Scales every element             |
| **Comparison**             | `a > 5`        | Returns boolean array            |
| **Mathematical Functions** | `np.sqrt(a)`   | Applies square root element-wise |
| **Dot Product**            | `np.dot(a, b)` | Matrix multiplication            |


Example:

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a + b)
print(a * b)
print(np.sqrt(a))
```

### **🔹 Performance Comparison**

```python
import numpy as np, time

x = np.arange(1_000_000)

# Vectorized
start = time.time()
y = x * 2
print("NumPy time:", time.time() - start)

# Loop-based
lst = list(range(1_000_000))
start = time.time()
y = [i * 2 for i in lst]
print("List time:", time.time() - start)
```

### **✅ Summary**

A vectorized operation in NumPy means performing operations on entire arrays at once — without explicit loops.
It makes numerical computations faster, cleaner, and more memory-efficient, which is why NumPy is essential for data science, machine learning, and scientific computing.

# **Question 10  What is the significance of hierarchical indexing in Pandas ?**

Answer:

🔹 Definition

Hierarchical indexing means using more than one index level on rows or columns to organize and access data more efficiently.

It helps manage and analyze data with multiple keys, like data grouped by Region → City → Year or Department → Employee.

## **🔹 Example: Creating a MultiIndex**



```python
import pandas as pd

# Define data
data = {
    'Sales': [250, 300, 400, 500],
    'Profit': [25, 35, 45, 55]
}

# Create hierarchical index
index = pd.MultiIndex.from_tuples(
    [('East', 'Delhi'), ('East', 'Kolkata'),
     ('West', 'Mumbai'), ('West', 'Pune')],
    names=['Region', 'City']
)

df = pd.DataFrame(data, index=index)
print(df)
```

## **🔹 Why Hierarchical Indexing Is Useful**

| **Advantage**                      | **Explanation**                                                                               |
| ---------------------------------- | --------------------------------------------------------------------------------------------- |
| **1. Better data organization**    | Helps represent complex, multi-level data clearly in a 2D table.                              |
| **2. Easy subsetting and slicing** | You can access data at different levels (like all cities in a region).                        |
| **3. Easier aggregation**          | Ideal for group-based operations (like `groupby()` results).                                  |
| **4. Pivot-like representation**   | Enables reshaping data easily with `stack()` and `unstack()`.                                 |
| **5. Hierarchical columns**        | Works with columns too — useful for multi-dimensional metrics (e.g., Sales & Profit by Year). |



### **🔹 Accessing Data in a MultiIndex**
Access all data for one region

```python
print(df.loc['East'])
```

### **Access a single value**

```python
print(df.loc[('East', 'Delhi'), 'Profit'])


```

### **Access all profits from the “West” region**



```python
print(df.loc['West', 'Profit'])
```

### **🔹 Working with Columns as MultiIndex**


```python
arrays = [
    ['Sales', 'Sales', 'Profit', 'Profit'],
    ['2023', '2024', '2023', '2024']
]
columns = pd.MultiIndex.from_arrays(arrays, names=('Metric', 'Year'))

df = pd.DataFrame([[200, 250, 30, 40],
                   [180, 300, 25, 45]], columns=columns)
print(df)
```

### **🔹 Common Methods**

| Method            | Purpose                            |
| ----------------- | ---------------------------------- |
| `df.index.names`  | View or set index level names      |
| `df.swaplevel()`  | Swap index levels                  |
| `df.sort_index()` | Sort by index                      |
| `df.unstack()`    | Convert one index level to columns |
| `df.stack()`      | Convert columns back to an index   |


### **🔹 When to Use Hierarchical Indexing**

    * When working with multi-dimensional or grouped data
    * After using groupby(), which naturally returns a MultiIndex
    * When reshaping data using pivot_table(), stack(), or unstack()
    * For time-series, regional, or categorical breakdowns

###*✅ In Summary*

| **Feature**     | **Explanation**                                                |
| --------------- | -------------------------------------------------------------- |
| **Concept**     | Multi-level indexing for rows and/or columns                   |
| **Purpose**     | Represent and access high-dimensional data in 2D               |
| **Key Benefit** | Easier grouping, slicing, reshaping, and aggregation           |
| **Typical Use** | Results of `groupby()`, pivot tables, and multi-level datasets |

# **Question 12  What is the purpose of the describe() function in Pandas?**

# **Question 17  Explain the difference between apply() and map() in Pandas ?**

Answer:

In Pandas, both apply() and map() are used to apply functions to data — but they work at different levels and have different scopes. Here’s a clear breakdown:

🔹 map()

* Used with: Series (one-dimensional)
* Purpose: Applies a function element-wise to each value in the Series.
* Scope: Works only on one column (or Series) at a time.

Syntax:

```python
Series.map(function)
```

Example:

```python
import pandas as pd

s = pd.Series([1, 2, 3, 4])
s.map(lambda x: x * 2)
```

✅ You can also use a dictionary or Series with map() for value mapping:

```python
s.map({1: 'A', 2: 'B', 3: 'C'})
```

🔹 apply()

Used with: Both DataFrame and Series

* Purpose: Applies a function along an axis (rows or columns) of a DataFrame or element-wise on a Series.

* Scope: Can work on multiple columns or rows at once.

Syntax:

```python
DataFrame.apply(function, axis=0 or 1)
```

axis=0 → apply function to each column

axis=1 → apply function to each row

Example:

```python
df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [10, 20, 30]
})

# Apply function to each row
df.apply(lambda x: x['A'] + x['B'], axis=1)
```

| Axis parameter               | ❌ No                           | ✅ Yes                                                      |

# **Question 21  Why is NumPy’s array slicing faster than Python’s list slicing?**
Answer:
NumPy’s array slicing is faster than Python’s list slicing because of how data is stored, managed, and accessed in memory.

## **🔹 1. Contiguous Memory Storage**

* NumPy arrays are stored in contiguous blocks of memory (just like in C or Fortran).

* Python lists are arrays of pointers — each element can be stored anywhere in memory.

✅ Effect:
NumPy can access elements directly using pointer arithmetic, while lists must follow references one by one.

Illustration:

```python
NumPy array: [1, 2, 3, 4]  → stored as one block of numbers
Python list: [ptr1, ptr2, ptr3, ptr4] → each points to a separate object
```

## **🔹 2. Views vs. Copies**

* When you slice a NumPy array, it usually creates a view of the same underlying data — no new memory is allocated.

* When you slice a Python list, it creates a new list (a full copy of the data).

✅ Effect:
NumPy slicing = zero-copy operation
Python list slicing = copying data to a new list

Example:

```python
import numpy as np

arr = np.arange(10)
sub = arr[2:6]
sub[0] = 999
print(arr)  # Original changes!
```

→ Because sub is a view, not a copy.

In contrast:

```python
lst = list(range(10))
sub = lst[2:6]
sub[0] = 999
print(lst)  # Original unchanged
```

## **🔹 3. Fixed Data Types**

* NumPy arrays have a single, fixed data type (e.g., float64, int32).

* Python lists can hold mixed data types (e.g., int, str, float, object).

✅ Effect:
NumPy can perform vectorized operations directly on the raw memory buffer, without type-checking each element like Python lists must do.

## **🔹 4. Low-Level (C-based) Implementation**

* NumPy’s slicing, indexing, and arithmetic are implemented in compiled C code, not Python.

* This means minimal interpreter overhead and much faster execution.

✅ Effect:
NumPy slicing and operations are orders of magnitude faster than Python’s interpreted loops or list slicing.

## **🔹 5. Cache Efficiency**

Because NumPy arrays are contiguous and homogeneous, they are CPU-cache-friendly — sequential access takes advantage of spatial locality.

✅ Effect:
Data retrieval and slicing are much faster due to efficient use of CPU cache lines.

## **🔹 6. Vectorized Computations (Bonus)**

NumPy supports vectorized operations — slicing and operating on entire blocks of data at once — while Python lists require explicit loops.

Example:

```python
import numpy as np

a = np.arange(1e6)
b = a[1000:2000] + 10   # Vectorized
```

Equivalent Python list operation would require:

```python
lst = list(range(int(1e6)))
b = [x + 10 for x in lst[1000:2000]]  # Loop-based
```

### **✅ Conclusion:**

NumPy array slicing is faster because it operates on contiguous, typed, C-level memory blocks, creating views instead of copies, while Python list slicing must copy and reference individual Python objects one by one.

# **Question 22 What are some common use cases for Seaborn?**

Answer:
Seaborn, a Python library built on Matplotlib, is excellent for statistical data visualization. Its common use cases center around Exploratory Data Analysis (EDA), particularly when working with complex datasets in a structured format like a Pandas DataFrame.


Here are some of the most common use cases for Seaborn:

## **📊 Visualizing Distributions**
Seaborn excels at understanding the spread and shape of a single variable or comparing distributions across different groups.

* Histograms and Kernel Density Estimates (histplot, kdeplot, displot): Used to visualize the probability distribution of a continuous variable.

* Box Plots (boxplot): Quickly summarizing the distribution (median, quartiles, and outliers) of a variable, especially useful for comparing distributions across different categories.


* Violin Plots (violinplot): Similar to box plots but providing a richer, smoothed density estimate of the data distribution.

## ** Exploring Relationships**
Seaborn makes it easy to visualize the relationships between two or more variables.

* Scatter Plots (scatterplot): The primary tool for showing the relationship between two continuous variables. You can easily add a third categorical variable using color (hue).


* Line Plots (lineplot): Ideal for visualizing trends, particularly over time or another continuous dimension.

* Heatmaps (heatmap): Excellent for visualizing matrix-like data, such as correlation matrices (to see how all pairs of variables relate) or pivot tables.


* Regression Plots (regplot, lmplot): Visualizing scatter plots and automatically fitting and plotting a linear regression model and its confidence interval, helping to model linear relationships.

## **Working with Categorical Data**
It provides specialized tools for comparing a numeric variable across different categories.

* Bar Plots (barplot): Used to show the central tendency (e.g., mean or median) and confidence intervals of a numeric variable for distinct categories.


* Count Plots (countplot): Showing the number of observations in each categorical bin, essentially a histogram for a categorical variable.


* Swarm Plots (swarmplot): Drawing non-overlapping scatter plots to show the distribution of a numeric variable within categories where all points are visible.

## **Creating Multi-Plot Structures**
Seaborn simplifies creating complex figures containing multiple plots to view different subsets or relationships simultaneously.

* Pair Plots (pairplot): Creating a grid of plots where all numerical variables in a dataset are plotted against each other (scatter plots) and their individual distributions are shown on the diagonal. This is a crucial first step for multivariate analysis.

* Joint Plots (jointplot): Combining two plots, such as a scatter plot and a histogram, to show both the bivariate relationship and the univariate distributions on the same figure.

* Facet Grids (FacetGrid, relplot, catplot, displot): Structuring a dataset into smaller subsets and plotting the same type of visualization for each subset (e.g., showing a separate scatter plot for each day of the week).

Creating a 2D NumPy array and calculating the sum of each row is straightforward using the NumPy library.

## **Creating a 2D NumPy Array**
You typically create a 2D array by passing a Python list of lists to the np.array() function. Each inner list represents a row in the 2D array.



```python
import numpy as np

# A list of lists representing a 3x4 matrix
data = [
    [10, 20, 30, 40],
    [5, 15, 25, 35],
    [1, 2, 3, 4]
]

# Create the 2D NumPy array
two_d_array = np.array(data)

print("2D Array:")
print(two_d_array)
print("\nShape (Rows, Columns):", two_d_array.shape)
```

## **Calculating the Sum of Each Row**
To calculate the sum of each row, you use the sum() method (or the np.sum() function) and specify the axis argument.

* In NumPy, axis=1 refers to the dimension across the columns, which calculates a summary statistic for each row.

* Conversely, axis=0 refers to the dimension across the rows, which would calculate a summary statistic for each column.

```python
# Calculate the sum of each row (axis=1)
row_sums = two_d_array.sum(axis=1)

print("\nSum of Each Row (axis=1):")
print(row_sums)
```

# **Question 4. How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap?**
Answer:

The workflow involves two main steps:

1. Calculate the Correlation Matrix: Use the .corr() method from Pandas on your DataFrame.

2. Visualize the Matrix: Use Seaborn's heatmap() function to display the matrix with colors.

I will generate a single Python script that demonstrates this using a sample dataset.

## **Key Components:**
* correlation_matrix = numerical_df.corr(): This Pandas method automatically calculates the Pearson correlation coefficient between all pairs of numeric columns.

* sns.heatmap(...):

    * correlation_matrix is passed as the main data.

    * annot=True is crucial as it displays the actual correlation numbers (e.g., 0.98) inside the cells, which helps in precise interpretation.

    * cmap='coolwarm' uses blue for negative correlations, red for positive, and white/gray for zero correlation, which is the standard color scheme for this type of visualization.

## **🔹 Step 1: Import Required Libraries**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## **🔹 Step 2: Load a Dataset**

You can use any dataset — let’s use Seaborn’s built-in tips dataset for example:


```python
df = sns.load_dataset("tips")
```

## **🔹 Step 3: Calculate the Correlation Matrix**

The correlation matrix measures the linear relationship between numerical columns:

```python
corr = df.corr(numeric_only=True)
```

## **🔹 Step 4: Visualize with a Heatmap**

Now, use Seaborn’s heatmap() to visualize it:

```python
plt.figure(figsize=(8, 6))
sns.heatmap(corr,
            annot=True,          # display correlation values
            cmap='coolwarm',     # color palette
            fmt=".2f",           # format to 2 decimal places
            linewidths=0.5,      # add lines between cells
            square=True)         # make cells square
plt.title("Correlation Matrix Heatmap")
plt.show()
```

## **✅ Explanation**

* corr() → calculates pairwise correlation (default method = Pearson).
* annot=True → shows numeric values inside the heatmap cells.
* cmap → defines color theme (e.g., "coolwarm", "viridis", "RdBu").
* Dark red/blue areas indicate strong correlation (positive/negative).
* Values close to 0 mean weak or no correlation.

# **Question 13  Create a 3D scatter plot using Plotly ?**

```python
import plotly.express as px
import pandas as pd

# Sample data
data = {
    'X': [1, 2, 3, 4, 5],
    'Y': [10, 11, 12, 13, 14],
    'Z': [5, 6, 7, 8, 9],
    'Category': ['A', 'B', 'A', 'B', 'A']
}

df = pd.DataFrame(data)

# Create 3D scatter plot
fig = px.scatter_3d(
    df,
    x='X',
    y='Y',
    z='Z',
    color='Category',   # Color by category
    size='Z',           # Bubble size based on Z values
    symbol='Category',  # Different symbols for each category
    title='3D Scatter Plot using Plotly'
)

# Show the figure
fig.show()
```
